```python
nulls = df.isna().sum().sort_values(ascending=False)
nulls[nulls > 0]
```

```python
duplicados = df.duplicated().sum()
duplicados
```

```python
df["rating"].describe()
```

```python
top_categorias = df["category"].value_counts().head(10)
top_categorias
```

```python
plt.figure(figsize=(10, 5))
df["rating"].hist(bins=30)
plt.title("Distribuição de rating")
plt.xlabel("rating")
plt.ylabel("frequência")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(x=top_categorias.values, y=top_categorias.index)
plt.title("Top 10 categorias")
plt.xlabel("Quantidade")
plt.ylabel("Categoria")
plt.show()
```
